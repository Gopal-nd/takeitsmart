```python
vector database is a database which is designed to store,to search vector embeddings
which are numerical representations of the data like text,images,audios and videos
```

```python
vector is nothing it is used to convert the raw data into numbers(vectors)
```

```python
text="i love nlp"--->[0.12,-0.67,0.34---]

these vectors are helpful to capture semantic meaning
```

```python
in traditional data base like sql:
excat match
filters

vector db solves ny using:
similarity search
```

```python
whenever we give any query vector,the db finds the closest vectors

common metrics:

cosine similarity
euclidean distance
dot product
```

```python
Architecture of vector db:

1.input data like text,images,audio and videos
2.embedding model:which helps to convert the textual data into vectors
3.storage:stores the vectors in vector database
4.indexing:which helps for efficient search
5.query processing:convert query--->vector---->search nearest neighbors
```

```python
popular vector databases:
faiss(facebook ai similarity search)
pinecone(cloud based)
chromadb
milvus
vespa
weaviate
qdrant and many more
```

```python
RAG:(retrieval augumented generation):mainly used with llm like chatgpt

user query--->embedding
retrieve relevant docs from the vector db
feed to the llm--->generate the response

recommendation systems:

netflix,amazon etc
```
