```python
langchain:
it is an open source framework which is used for building web application by large language
models like gpt

in simple terms,langchain helps to connect an ai model to the real world data tools and also
workflows which helps to build the real world application like building the agents,
documents,qa systems
```

```python
langchain can also build multi step intelligence process:
the user will asks a question
searches the documents
extract the relevant information
generate the answer
```

```python
key components of langchain:

1.chains:A sequence of steps combining the prompts and also model calls

user input--->prompt template--->llm---->formatted output

2.Agents:
agents decide what action to take next

it can take the help of different tools:
web search
apis
websites
databases

3.memory:

this step helps to remember the previous information

4.rag:(retrieval augumented generation):

connects the llm with our own data(pdfs,databases,word etc)

examples:

store documents in a vector database
retrieve the relevant chunks
feed them to the model

5.tools & integration:
langchain plug into llms,databases and external sources
```

```python
query|llm|answer
```

```python
from langchain.chat_models import chatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain.chains import LLMChain
llm=ChatOpenAI()
prompt=ChatPromptTemplate.from_template("Explain {topic} simple way")
chain=LLMchain(llm=llm,prompt=prompt)
print(chain.run("Langchain"))
```

```python
llama index:

it is one of the framework which is designed specifically to connect with large language models with
our own data mainly it is used in the rag applications

load--->structure---->index--->query(data using llms)
```

```python
workflow:

1.data ingestion:

load the data from text,pdfs,csv,word,databases etc


2.indexing:

transform the data into searchable structure

types of index:

vector index(commonly used)
tree index
keyword index


3.retrieval:

when the user asks a question

it finds the relevant chunks from our data

4.response generation:

send the retrieval context to the llm

generate the accurate answers
```

```python
from llama_index.core import VectorStoreIndex,SimpleDirectoryReader

documents=SimpleDirectoryReader("data").load_data()

index=VectorStoreIndex.from_documents(documents)

query_engine=index.as_query_engine()

response=query_engine.query("what is langchain")

print(response)
```

```python
Lang graph:

it is a framework which is used for multi step ai based flow

it is mainly used to build the agents

it mainly consists of node,branch,loop

mainly it helps to solve building agents for automation
```

```python
key components of lang graph:

1.state:
it helps to share memory across nodes

2.node:
functions that performs tasks

3.edges:
it defines the flow between the nodes

4.graph:
it helps to combine nodes+edges
```

```python
langsmith:

it is a developer platform for debugging,testing and monitoring the llm applications built with
tools like langchain and langgraph

langchain---build apps
langgraph---control workflow
langsmith---observe,debug, and helps to improve them
```
