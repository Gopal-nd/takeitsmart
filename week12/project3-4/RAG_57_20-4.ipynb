```python
Rag(retrieval augumented generation):
it is the architecture where we get the:
relevant information that can be retrieved from an external knowledge this
information can be augumented into the prompt
so that the llm can generate the final answer
```

```python
rag is nothing but it allows the llm models to answer the question/prompt from the external
documents not just by their trained memory
```

```python
what is the leave policy in our company

A normal llm says:

company have 11 paid leaves----hallicunations

with the help of rag:

1.search the hr documents
2.find the leave policy paragraph
3.send to the llm model
4.generate the response


final answer:

employees are entitled to 20 annual paid leaves
```

```python
why this rag is important:

because the llms like  chatgpt or grok doesnot have the access to the private information
it only have the access to internet data which was it trained on
it cannot access the real time internal data

it may hallicinate the answers so we require rag
```

```python
why rag is needed?
because to avoid the hallucinations

problem 1: static knowledge:

llms are trained once it consists of only old data

whenever the user asks like:

what is our latest hr policy?

it does not how to answer

problem 2: hallucinations:

llms some times generate wrong answers with confidently

how many leave days in our company

without context:

employees get 15 paid leaves annually

but actual output is 20

this is hallucination


problem 3:private data:

llms do not know:company pdfs,internal documents,research papers,private db records


this is why rag is needed to solve all the above problems

with the help of rag we can easily:
retrieve real data
giving it to the llm model 
and generating the real answers from the actual documents
```

```python
Architecture of rag:

1.data ingestion:

this is used to collect the data from pdfs,word files,csvs,databases,apis,websites

examples:

company policy documents
product manuals
employees details

why ingestion is required:

because the llm needs custom knowledge base

without data the llm cannot retrieve the information

example:

pdf contains:

employees are entitled to 20 leaves annually

this is the raw input which we can take
```

```python
2.chunking:

large documents are splitting into small chunks

why?

because the llm has token limits
retrieval works better on smaller pieces

example:

employees are entitled to 20 annual leaves
maternity leave is 90 days
sick leaves are 10 days
monday to thursday casual wear
references are allowed

split into chunks:

chunk1:
employees are entitled to 20 annual leaves

chunk2:
maternity leave is 90 days

chunk3:
sick leaves are 10 days
```

```python
3.embeddings:

it is used to convert the text data  into vectors

why embeddings are important:

because the machines cannot understand textual data directly so conver to numbers(vectors)

some of the popular models:
openai embeddings
sentence transformers
```

```python
4.vector db:
it is used to store the data into  vectors

different vector databases like
faiss
pinecone
chroma db
weaviate
```

```python
5.user query embedding:

user:how many annual leaves

this query convert into embeddings and stores in the vector data base
```

```python
6.similarity search/retrieval:

the retrieval finds the most relevant chunks

using:
cosine similarity
euclidean distance 
```

```python
7.prompt augumentation:
now the retrieved chunks is added to the prompt

prompt becomes:

context:
employees are entitled to 20 annual leaves

question:
how many leaves do employee get?
```

```python
8.llm generation:

llm reads user question
to retrieve context

"employees get 20 annual leaves"

this helps to reduce the hallucinations
```

```python
Advance rag:
1.re-ranking:

it helps to improve the quality of the retrieved results before giving to the llm model

deeper understanding and to get the accurate results:

user query--->vector retrieval--->re ranking--->top n(best chunks)--->llm
```

```python
2.hybrid search:

keyword search+hybrid search

keyword search:it gives the excat matching
hybrid search--semantic understanding
```

```python
user query--->keyword search--->vector search--->combine both results--->re-ranking--->top results
```

```python
3.meta data filtering:

it is the extra information attached to each document/chunk that helps in filtering the data

user query--->metadata filter--->retrieved from filtered data--->re ranking--->llm
```
