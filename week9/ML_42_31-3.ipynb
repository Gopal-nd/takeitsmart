```python
knn:
it is used to classify the new data based on the majority of the voting with the help of euclidean distance formula

choose the k value(eg:3 or 5)
calculate the eucldiean distance formula
pick the k nearest values
take the majority of the voting
finally assign which class it belongs to
```

```python
svm(support vector machine)
it is one of the ml algorithm which is used to find the best boundary(hyperplane)
between the classes

w.x+b=0
```

```python
naive bayes:
it is a probabilistic classification based on the theorem(bayes)
p(a/b)=p(b/a).p(a)/p(b)

types of naive bayes:
1.gaussian(continuous data)
2.multinomial(for text data)
3.bernoulli(binary features)
```

```python
decision tree:
it uses tree like structure to make decisions 
it consists of node,branches and leaf
it follows asm(Attribute selection measure):
1.information gain
2.gini index

it faces the overfitting problem to reduce this we have random forest
```

```python
random forest:
it is also called as ensemble learning
it consists of bagging and boosting techniques
it have multiple decision trees
```

```python
from sklearn.datasets import load_iris
```

```python
data=load_iris()
```

```python
print(data.feature_names)
```

```python
print(data.target_names)
```

```python
x=data.data
```

```python
y=data.target
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
acc=accuracy_score(y_test,y_pred)
```

```python
acc
```

```python
classification_report(y_test,y_pred)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model=KNeighborsClassifier(n_neighbors=3)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.svm import SVC
```

```python
model=SVC()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
model=AdaBoostClassifier()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```
