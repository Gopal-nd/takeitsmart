```python
deep learning:
it is used to work with neural networks
it is the way to understand the machines to learn and to transform the data so that it can easily
identify the patterns inside it
it mimics like a human brain
```

```python
ml:input data---->feature extraction(manually)--->algorithm--->output
dl:input data---->feature extraction(automatically)--->algorithm--->output
```

```python
input layer---->hidden layer---->output layer
```

```python
features of deep learning:
feature extraction can be done automatically
it completely works on the neural networks
it works on huge amount of the data
```

```python
Applications of deep learning:
speech recognition
self driving cars
anamoly detection
image classification
nlp tasks
fraud detection
```

```python
implementation of deep learning:
1.data collection/data preprocessing
text,images---clean,train,test,split
2.model architecture:
choose the model(cnn,rnn,lstm,gru,gans,fnn,vaes,transformers)
3.forward propagation:
moving forwards from input--->hidden--->output
4.loss function(the error rate between actual value and predicted value)
5.backward propagation(adjusts the weights)
6.optimization(improving the model performance)
7.model evaluation
```

```python
Activation functions:
1.relu(rectified linear unit) f(x)=max(0,x)
2.sigmoid f(x)=1/1+e**-x
3.tanh (-1 to +1)---hyperbolic tangent
4.softmax(multi class classification)
```

```python
types of gradient descent:
1.batch gradient descent
2.mini batch gradient descent
3.sgd
4.adam
5.rmsprops
```

```python
Batch Gradient Descent is more accurate but slower and computationally 
expensive. It is ideal when working with small to medium-sized datasets
and when high accuracy is required.

Stochastic Gradient Descent, on the other hand, is faster and requires less 
computational power, making it suitable for large datasets. It can also escape
local minima more easily but may converge less accurately.
```

```python
types of deep learning:
1.feed forward neural networks(fnns/ann)----simple tasks like classification/regression
2.convolutional neural networks(cnn)---images
3.recurrent neural networks(rnn)----textual data
4.long short term memory(lstm)----textual data
5.gated recurrent units(gru)---text data
6.generative adversarial neural networks(gans)---images
7.variational auto encoders(vaes)---reducing dimensions
8.transformers---advance version of nlp(models like bert,roberta,t5,gpt etc)
```
