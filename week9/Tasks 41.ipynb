## SIMPLE LINEAR REGRESSION

## Dataset 1: Distance vs Fuel Consumption

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

data = {
    "Distance": [10, 20, 30, 40, 50, 60, 70],
    "Fuel": [1, 2, 3, 4, 5, 6, 7]
}

df = pd.DataFrame(data)

x = df[['Distance']]
y = df[['Fuel']]

model = LinearRegression()
model.fit(x, y)

y_pred = model.predict(x)

print(y_pred)

print("R2 Score:", r2_score(y, y_pred))
```

## MULTIPLE LINEAR REGRESSION

## Dataset 1: Student Performance

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

data = {
    "Hours": [2, 4, 6, 8, 10],
    "Sleep": [6, 7, 8, 6, 7],
    "Marks": [50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)

x = df[['Hours', 'Sleep']]
y = df[['Marks']]

# Model
model = LinearRegression()
model.fit(x, y)

# Predictions
y_pred = model.predict(x)

# Accuracy
print("R2 Score:", r2_score(y, y_pred))

# New prediction
print("Prediction:", 
      model.predict(pd.DataFrame([[7, 7]], columns=['Hours', 'Sleep'])))
```

## Dataset 2: Car Price Prediction

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

data = {
    "Engine": [1000, 1200, 1500, 1800, 2000],
    "Mileage": [20, 18, 15, 12, 10],
    "Price": [300000, 350000, 500000, 700000, 900000]
}

df = pd.DataFrame(data)

x = df[['Engine', 'Mileage']]
y = df[['Price']]

# Split data 
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.4, random_state=42)

# Model
model = LinearRegression()
model.fit(x_train, y_train)

# Predictions on test data
y_pred = model.predict(x_test)

# Metrics
print("R2 Score:", r2_score(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))

# New prediction 
print("Prediction:", 
      model.predict(pd.DataFrame([[1600, 14]], columns=['Engine', 'Mileage'])))
```

## Dataset 3: Employee Productivity

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

data = {
    "Hours_Worked": [4, 6, 8, 5, 7, 9, 10, 3],
    "Experience": [1, 2, 3, 2, 3, 4, 5, 1],
    "Productivity": [40, 55, 70, 50, 65, 80, 90, 35]
}

df = pd.DataFrame(data)

x = df[['Hours_Worked', 'Experience']]
y = df[['Productivity']]

# Split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=42
)

# Model
model = LinearRegression()
model.fit(x_train, y_train)

# Prediction on test data
y_pred = model.predict(x_test)

# Metrics
print("R2 Score:", r2_score(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))

# New prediction
print("Prediction:", 
      model.predict(pd.DataFrame([[8, 3]], columns=['Hours_Worked', 'Experience'])))
```

## POLYNOMIAL REGRESSION

## Dataset: Speed vs Stopping Distance

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures

speed = np.array([10, 20, 30, 40, 50, 60]).reshape(-1,1)
distance = np.array([5, 20, 45, 80, 125, 180])

# Linear model
lin_model = LinearRegression()
lin_model.fit(speed, distance)
lin_pred = lin_model.predict(speed)

# Polynomial model
poly = PolynomialFeatures(degree=2)
speed_poly = poly.fit_transform(speed)

poly_model = LinearRegression()
poly_model.fit(speed_poly, distance)
poly_pred = poly_model.predict(speed_poly)

print("Linear R2:", r2_score(distance, lin_pred))
print("Polynomial R2:", r2_score(distance, poly_pred))
```
