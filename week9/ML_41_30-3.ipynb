## Multi linear regression

```python
np.random.seed(42)
n=100
size=np.random.randint(500,4000,n)
bedrooms=np.random.randint(1,6,n)
age=np.random.randint(1,10,n)
price=(50000+(size*150)+(bedrooms*10000)+(age*12000)+np.random.normal(0,20000,n))
```

## Example 1

```python
data=pd.DataFrame({
    'Size':size,
    'Bedrooms': bedrooms,
    'Age': age,
    'Price': price
})
```

```python
data
```

```python
x=data[['SQFT','BEDROOMS','AGE']]
```

```python
y=data[['PRICE']]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```

## Example 2

```python
#import the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.datasets import fetch_california_housing
```

```python
housing_data=fetch_california_housing()
df=pd.DataFrame(housing_data.data,columns=housing_data.feature_names)
```

```python
housing_data
```

```python
df
```

```python
df['MedHouseVal']=housing_data.target
```

```python
req_features=['MedInc','AveRooms','AveOccup']
x=df[req_features] #independent features
y=df['MedHouseVal'] #dependent feature
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
print("MSE -- MEAN SQUARE ERROR:", mean_squared_error(y_test,y_pred))
print("R^2 Score:", r2_score(y_test,y_pred))
```

## Polynomial regression

## Example 1

```python
import numpy as np
x=np.array([1,2,3,4,5,6])
```

```python
x
```

```python
x=np.array([1,2,3,4,5,6]).reshape(-1,1)
```

```python
x
```

```python
y=np.array([1,4,9,16,25,36])
```

```python
y
```

```python
plt.plot(x,y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_poly,y)
```

```python
y_pred=model.predict(x_poly)
```

```python
y_pred
```

```python
r2_score(y,y_pred)
```
